```python
# Part B – Question 5
# Monthly Electricity Consumption Trend Analysis Using Python

# Create or download a CSV file containing at least 12 months of electricity consumption records.
# Read and inspect the dataset using Pandas.
# Handle missing and inconsistent values.
# Convert the Date column into the appropriate date format.
# Analyse monthly Electricity Consumption.
# Create a line graph showing electricity consumption over time.
# Add title, axis labels, date formatting, line style, markers and grid.
# Identify months with the highest and lowest electricity consumption.
# Analyse seasonal trends and sudden increases or decreases.
# Save the graph as an image and prepare a brief report.

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

data = pd.read_csv("electricity.csv")
sns.lineplot(
    x="Date",
    y="Electricity Consumption",
    hue="Consumer Category",
    linestyle="--",
    style="Consumer Category",
    marker="o",
    data=data
)

plt.xlabel("Time Lines")
plt.ylabel("Electricity Consumption (kWh)")
plt.title("Monthly Electricity Consumption Trend")

plt.xticks(rotation=45, ha="center", fontsize=12,
           fontfamily="serif", fontweight="bold")

plt.yticks(rotation=45, ha="center", fontsize=12,
           fontfamily="serif", fontweight="bold")

plt.grid(True)
plt.show()
```
