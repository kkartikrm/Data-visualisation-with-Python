```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("p5.csv")
sns.set_style("whitegrid")
data = data.dropna()

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=data,
    x="Date",
    y="Closing Price",
    marker="o",
    color="green",
    label="Closing Price"
)

plt.title("Stock Market Trend Analysis")
plt.xlabel("Dates")
plt.ylabel("Closing Price")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.savefig("stock_market_trend.png")
plt.show()

print("Maximum Closing Price:", data["Closing Price"].max())
print("Minimum Closing Price:", data["Closing Price"].min())
print(data.head())
```
