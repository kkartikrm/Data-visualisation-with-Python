```python
# Part B – Question 6
# Website Traffic Analysis Using Interactive Plotly Visualization

# Create or download a dataset containing at least 30 website records.
# Read and inspect the dataset using Pandas.
# Handle missing and inconsistent values.
# Convert the Date column into date format.
# Create an interactive line plot to visualize website visitors over time.
# Add a suitable title and axis labels.
# Display date and visitor count using hover information.
# Customize colours, markers, templates and gridlines.
# Enable zooming and panning.
# Identify the highest and lowest website traffic.
# Analyse traffic trends, spikes and drops.
# Save the interactive graph as an HTML file.
# Prepare a brief report summarizing the observations.

import pandas as pd
import plotly.express as px

df = pd.read_csv("website_traffic.csv")

df["Date"] = pd.to_datetime(df["Date"])

fig = px.line(
    df,
    x="Date",
    y="Visitors",
    title="Website Traffic Analysis",
    hover_data="Visitors"
)

fig.update_traces(
    mode="lines+markers",
    line=dict(color="blue", width=4),
    marker=dict(color="red", size=10)
)

fig.update_xaxes(tickangle=45)

fig.update_layout(
    xaxis_title="Date",
    yaxis_title="Number of Visitors",
    template="plotly_white"
)
fig.write_html("Website.png")
fig.show()
```
