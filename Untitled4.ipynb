```python
# Part B – Question 4
# Product Price Analysis Across Categories Using Python

# Create a CSV file containing at least 100 product records.
# The dataset should contain the following attributes:
# Product ID, Product Category, Product Name,
# Brand, Price and Customer Rating.

# Import the CSV dataset using Pandas.
# Inspect and understand the dataset.
# Handle missing and inconsistent values.
# Calculate summary statistics for product prices.

# Create a box plot to compare product prices
# across different product categories.

# Create a violin plot to visualize the distribution
# of product prices across different categories.

# Add suitable titles and axis labels.
# Customize the colours and plot theme.
# Rotate category labels for better readability.

# Analyse the highest and lowest median prices.
# Identify price variations and outliers.
# Compare price distributions across categories.

# Save the graphs as image files.
# Prepare a brief report based on the analysis.


import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("kar.csv")

print("First 5 Records:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

df["Product Category"] = df["Product Category"].astype("string").str.strip().str.title()
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
df["Customer Rating"] = pd.to_numeric(df["Customer Rating"], errors="coerce")

df["Product Category"] = df["Product Category"].fillna("Unknown")
df["Product Name"] = df["Product Name"].fillna("Unknown")
df["Brand"] = df["Brand"].fillna("Unknown")
df["Customer Rating"] = df["Customer Rating"].fillna(df["Customer Rating"].median())

df.dropna(subset=["Price"], inplace=True)

print("\nPrice Summary Statistics:")
print(df["Price"].describe())

print("\nMedian Price by Category:")
median = df.groupby("Product Category")["Price"].median()
print(median)

print("\nPrice Variation by Category:")
variation = df.groupby("Product Category")["Price"].std()
print(variation)

print("\nHighest Median Price:")
print(median.idxmax(), median.max())

print("\nLowest Median Price:")
print(median.idxmin(), median.min())

print("\nOutlier Analysis:")

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(1, 2, figsize=(16, 7))

sns.boxplot(data=df, x="Product Category", y="Price", ax=axes[0])

axes[0].set_title("Product Price Distribution by Category")
axes[0].set_xlabel("Product Category")
axes[0].set_ylabel("Price")
axes[0].tick_params(axis="x", rotation=20)

sns.violinplot(data=df, x="Product Category", y="Price",
               inner="quart", cut=0, ax=axes[1])

axes[1].set_title("Product Price Density by Category")
axes[1].set_xlabel("Product Category")
axes[1].set_ylabel("Price")
axes[1].tick_params(axis="x", rotation=20)

plt.tight_layout()
plt.show()

```
