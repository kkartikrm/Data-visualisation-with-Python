```python
# Part B – Question 3
# Hospital Patient Statistics Dashboard Using Python Subplots

# Create a CSV file containing hospital patient records.
# The dataset should contain the following attributes:
# Month, Patients Admitted, Patients Discharged,
# Emergency Cases and Average Treatment Cost.

# Import the CSV dataset using Pandas.
# Inspect and understand the dataset.
# Handle missing and inconsistent values.
# Select suitable variables for analysis.

# Create two subplots using Matplotlib and Seaborn.
# Create a line plot showing monthly patient admissions.
# Create a bar chart comparing emergency cases across months.

# Add suitable titles and axis labels.
# Customize colours, legends and gridlines.
# Adjust the subplot layout for better visualization.

# Analyse the highest and lowest patient admissions.
# Identify trends in emergency cases.
# Analyse the relationship between patient admissions
# and emergency cases.

# Save the combined figure as an image.
# Prepare a brief report based on the analysis.


import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv("p9.csv")

print(df.head())
print(df.isnull().sum())

df.fillna(df.median(numeric_only=True), inplace=True)

sns.set_theme(style="whitegrid")

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Line Plot – Monthly Patient Admissions
sns.lineplot(
    data=df,
    x="Month",
    y="Patients Admitted",
    marker="o",
    color="blue",
    ax=ax[0]
)

ax[0].set_title("Monthly Patient Admissions")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Patients Admitted")

# Bar Plot – Monthly Emergency Cases
sns.barplot(
    data=df,
    x="Month",
    y="Emergency Cases",
    color="orange",
    ax=ax[1]
)

ax[1].set_title("Monthly Emergency Cases")
ax[1].set_xlabel("Month")
ax[1].set_ylabel("Emergency Cases")

plt.tight_layout()
plt.savefig("hospital_analysis.png")
plt.show()

print("Maximum Admissions:", df["Patients Admitted"].max())
print("Minimum Admissions:", df["Patients Admitted"].min())

print("Maximum Emergency Cases:", df["Emergency Cases"].max())
print("Minimum Emergency Cases:", df["Emergency Cases"].min())
```
