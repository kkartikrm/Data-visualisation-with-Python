```python
# Part B – Question 2
# Vehicle Performance Analysis Using Python and Seaborn

# Create or download a CSV file containing at least 50 vehicle records.
# Include Vehicle Model, Engine Capacity, Horsepower,
# Fuel Efficiency, Vehicle Weight and Price.
# Read and inspect the dataset using Pandas.
# Handle missing and inconsistent values.
# Select two numerical variables for analysis.
# Create a scatter plot using Seaborn.
# Add suitable title, X-axis label and Y-axis label.
# Customize marker size, colour and transparency.
# Include a legend if applicable.
# Identify positive, negative or no correlation.
# Identify clusters and outliers.
# Save the graph as an image and prepare a brief report.


import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv("vehicles.csv")

print(df.head())
print(df.isnull().sum())

df.fillna(df.median(numeric_only=True), inplace=True)

correlation = df["Engine Capacity"].corr(df["Fuel Efficiency"])

print("Correlation:", round(correlation, 2))

sns.scatterplot(
    data=df,
    x="Engine Capacity",
    y="Fuel Efficiency",
    hue="Horsepower",
    size="Vehicle Weight",
    sizes=(60, 180),
    alpha=0.8,
    palette="viridis",
    edgecolor="black"
)

plt.title("Vehicle Engine Capacity vs Fuel Efficiency")

plt.xlabel("Engine Capacity (CC)")
plt.ylabel("Fuel Efficiency (km/l)")

plt.xlim(700, 3300)
plt.ylim(5, 30)

plt.grid(True)
plt.tight_layout()

plt.savefig("vehicle-analysis.png", dpi=300)

plt.show()
```
